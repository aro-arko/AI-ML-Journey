```python
# generator
lst = [x for x in range(500)]

def data_loader(chunk_size, lst):
    for i in range(0, len(lst), chunk_size):
        yield lst[i: i + chunk_size]

print(type(data_loader(5, lst)))

x = data_loader(5, lst)
print(next(x))
print(next(x))
```

## D. The Model Registry

```python

def model_info(name, layers, lr=0.01):
    return f"Model: {name}, Layers: {layers}, LR: {lr}"


t = int(input())

for _ in range(t):
    line = input().split()
    request_type = int(line[0])
    name = line[1]
    layers = int(line[2])

    if request_type == 1:
        print(model_info(name, layers))

    elif request_type == 2:
        print(model_info(layers=layers, name=name))

    elif request_type == 3:
        lr = float(line[3])
        print(model_info(name, layers, lr))
```

## F. GPU Can Eat Only Three Images

```python
n, batch_size = map(int, input().split())

lst = [x for x in range(n+1)]

def batch_list(lst, batch_size):
    """
    Split a list into batches of a specified size.
    """
    for i in range(1, len(lst), batch_size):
        yield lst[i : i + batch_size]

for batch in batch_list(lst, batch_size):
    batch_mean = sum(batch) / len(batch)
    print(f"{batch} mean={batch_mean}")
```
